```python
import pandas as pd
import numpy as np
import glob
```

```python
# 배민
baemin_files = sorted(glob.glob("data/baemin/baemin*.csv"))

baemin_list = []

for file in baemin_files:
    df = pd.read_csv(file)
    baemin_list.append(df)

baemin = pd.concat(baemin_list, ignore_index=True)

baemin
```

```python
# 쿠팡
coupang_files = sorted(glob.glob("data/coupang/coupang*.csv"))

coupang_list = []

for file in coupang_files:
    df = pd.read_csv(file)
    coupang_list.append(df)

coupang = pd.concat(coupang_list, ignore_index=True)
coupang
```

```python
# cpc
baemin = baemin.rename(columns={
    "order_at": "주문일시",
    "total_bill": "전체금액",
    "ad_yn": "광고유입여부"
})

baemin["플랫폼"] = "배민"
baemin["광고집행비"] = baemin["cpc"]
baemin
```

```python
coupang = coupang.rename(columns={
    "order_time": "주문일시",
    "total_price": "전체금액",
    "advertisement_yn": "광고유입여부"
})

coupang["플랫폼"] = "쿠팡"
coupang["광고집행비"] = (
    coupang["전체금액"] * coupang["ad_pct"] / 100
)

coupang
```

```python
orders = pd.concat(
    [baemin, coupang],
    ignore_index=True
)

orders["주문일시"] = pd.to_datetime(orders["주문일시"])
orders["주문일자"] = orders["주문일시"].dt.date
orders = orders.sort_values("주문일시")
orders
```

```python
daily_sales = orders.pivot_table(
    index="주문일자",
    columns="플랫폼",
    values="전체금액",
    aggfunc="sum",
    fill_value=0
)

daily_sales["합계"] = daily_sales.sum(axis=1)
daily_sales
```

```python
ad_count = orders[orders["광고유입여부"] == "Y"].groupby("플랫폼").size()

total_count = orders.groupby("플랫폼").size()

ad_ratio = ad_count / total_count * 100

ad_cost = orders.groupby("플랫폼")["광고집행비"].sum()

ad_report = pd.DataFrame({
    "광고유입주문비율(%)": ad_ratio,
    "광고집행비": ad_cost
})

ad_report
```

```python
try:
    bank = pd.read_csv(
        "data/bank_20261001_20261031.csv",
        encoding="cp949"
    )
except UnicodeDecodeError:
    bank = pd.read_csv(
        "data/bank_20261001_20261031.csv",
        encoding="utf-8-sig"
    )

bank.head()
```

```python
food = bank[bank["거래처"] == "식자재마트"].copy()

food["거래일시"] = pd.to_datetime(food["거래일시"])

food["주문일자"] = food["거래일시"].dt.date

food_daily = food.groupby("주문일자")["출금액"].sum()

food_daily = food_daily.rename("식자재비")

food_daily
```

```python
result = daily_sales.merge(
    food_daily,
    left_index=True,
    right_index=True,
    how="left"
)

result["식자재비"] = result["식자재비"].fillna(0)
result["매출-식자재비"] = result["합계"] - result["식자재비"]
result = result.reset_index()

result = result[
    ["주문일자", "배민", "쿠팡", "합계", "식자재비", "매출-식자재비"]
]

result
```

```python
result.to_csv(
    "result.csv",
    index=False,
    encoding="utf-8-sig"
)
```
